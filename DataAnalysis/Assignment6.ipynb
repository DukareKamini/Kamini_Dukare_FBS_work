#### Dataset: The Cars93 dataset contains information about different car models, including their manufacturer, price, horsepower, MPG, and other attributes. You will use this dataset for the following tasks.  

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("Datasets/cars93.csv")
```

```python
data.head()
```

#### 1. Bar Chart: 

##### a. Create a bar chart to compare the average price of cars by manufacturer.

```python
avg_price = data.groupby('Manufacturer')['Price'].mean()
avg_price.plot(kind='bar', figsize=(10,6), color='blue')
plt.show()
```

##### b. Label the axes and add a suitable title.

```python
avg_price = data.groupby('Manufacturer')['Price'].mean()
avg_price.plot(kind='bar', figsize=(10,6), color='blue')
plt.xlabel('Manufacturer')
plt.ylabel('Average Price')
plt.title('Average Price of Cars by Manufacturer')
plt.show()
```

##### b. Use different colors for different manufacturers. 

```python
# Average price by manufacturer
avg_price = data.groupby('Manufacturer')['Price'].mean()

# Assign different colors automatically
colors = sns.color_palette('hsv', len(avg_price))  # unique color for each manufacturer

# Bar chart with different colors
plt.figure(figsize=(10,6))
sns.barplot(x=avg_price.index, y=avg_price.values, palette=colors)
plt.xlabel('Manufacturer')
plt.ylabel('Average Price')
plt.title('Average Price of Cars by Manufacturer')
plt.xticks(rotation=45)
plt.show()
```

#### 2. Line Chart: 

##### a. Create a line chart showing the trend of horsepower for different types of cars.

```python
sns.lineplot(x='Type', y='Horsepower', data=data, marker='o')
```

##### b. Include proper legends, labels, and a title 

```python
plt.figure(figsize=(10,6))
sns.lineplot(x='Type', y='Horsepower', data=data, marker='o')
plt.xlabel('Car Type')
plt.ylabel('Horsepower')
plt.title('Trend of Horsepower for Different Types of Cars')
plt.legend(['Horsepower'])
plt.show()
```

#### 3. Pie Chart: 

##### a. Create a pie chart showing the proportion of different car types in the dataset.

```python
data['Type'].value_counts().plot.pie()
```

##### b. Include percentages in the chart. 

```python
plt.figure(figsize=(8,8))
data['Type'].value_counts().plot.pie(autopct='%1.1f%%', colors=sns.color_palette('pastel'))
plt.title('Proportion of Different Car Types')
plt.ylabel('')  # Removes the default y-label
plt.show()
```

#### 4. Scatter Plot: 

##### a. Create a scatter plot to show the relationship between horsepower and MPG (Miles Per Gallon). 

```python
sns.scatterplot(x='Horsepower', y='MPG.city', data=data)
```

##### b. Label the axes and add a title.

```python
sns.scatterplot(x='Horsepower', y='MPG.city', data=data)
plt.xlabel('Horsepower')
plt.ylabel('MPG_city')
plt.title('Relationship between Horsepower and MPG_City')
plt.show()
```
